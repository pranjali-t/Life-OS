```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
apps = {
    "Instagram": "Social Media",
    "WhatsApp": "Social Media",
    "YouTube": "Entertainment",
    "Netflix": "Entertainment",
    "VS Code": "Coding",
    "Google Colab": "Coding",
    "Coursera": "Education",
    "Google": "Education",
    "Notion": "Productivity",
    "Spotify": "Entertainment"
}
```

```python
data = []

start_date = datetime(2026, 8, 1)

for i in range(14):
    date = (start_date + timedelta(days=i)).strftime("%Y-%m-%d")

    for app, category in apps.items():
        minutes = random.randint(15, 120)

        data.append([
            date,
            app,
            category,
            minutes
        ])
```

```python
df = pd.DataFrame(
    data,
    columns=["Date", "App_Name", "Category", "Minutes_Used"]
)

df.head(10)
```

```python
print("Total Rows:", len(df))
print("Total Days:", df["Date"].nunique())
print("Total Apps:", df["App_Name"].nunique())
```

```python
df.to_csv("screentime.csv", index=False)

print("✅ screentime.csv created successfully!")
```

```python
category_usage = (
    df.groupby("Category")["Minutes_Used"]
    .sum()
    .sort_values(ascending=False)
)

category_usage
```
